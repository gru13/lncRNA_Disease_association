```python
"""
Cell generated by Data Wrangler.
"""
def clean_data(df):
    # Sort by column: 'Algorithmic Value' (descending)
    df = df.sort_values(['Algorithmic Value'], ascending=[False])
    df["Ranking"] = range(1,len(df)+1)
    df = df[['Ranking'] + [col for col in df.columns if col != 'Ranking']]
    df['Normalized Algorithmic  Values'] = df['Algorithmic Value'].copy()
    
    # df.loc[df['Normalized Algorithmic  Values'] > 40, 'Normalized Algorithmic  Values'] = 40 + df['Normalized Algorithmic  Values'] * 1e-2
    df.loc[df['Normalized Algorithmic  Values'] < 3, 'Normalized Algorithmic  Values'] = 3 - df['Normalized Algorithmic  Values'] * 1e-1
    df.loc[df['Normalized Algorithmic  Values'] > 3, 'Normalized Algorithmic  Values'] = 3 + df['Normalized Algorithmic  Values'] * 1e-3
    df['Normalized Algorithmic  Values'] /= df['Normalized Algorithmic  Values'].max()

    return df

df_clean = clean_data(df.copy())
df_clean.head()
```
